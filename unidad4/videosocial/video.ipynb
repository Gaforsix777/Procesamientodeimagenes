Laplaciano  

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt 

img = cv2.imread('goku.jpg')
lap = cv2.Laplacian(img,cv2.CV_64F)

plt.imshow(lap,cmap = 'gray')
plt.xticks([]), plt.yticks([])
plt.title('laplaciano simple')
plt.show()

```

LAPLACIANO DEL GAUSSIANO

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt 

#suavizado de la imgaen 
blur = cv2.GaussianBlur(img,(5,5),0)

#laplaciano aplicando
lap = cv2.Laplacian(blur,cv2.CV_64F)

plt.imshow(lap,cmap = 'gray')
plt.xticks([]), plt.yticks([])
plt.title('laplaciano con suavizado')
plt.show()
```
